simpson 1/3 Gemini

```python
import math

# --- 1. นิยามฟังก์ชันสำหรับ Simpson's 1/3 Rule ---
def simpsons_one_third(func, a, b, n):
    """
    func: ฟังก์ชันที่ต้องการอินทิเกรต
    a: ขอบเขตล่าง
    b: ขอบเขตบน
    n: จำนวนช่วงย่อย (ต้องเป็นเลขคู่)
    """
    if n % 2 != 0:
        return "Error: n ต้องเป็นเลขคู่สำหรับ Simpson's 1/3 Rule"
    
    h = (b - a) / n
    
    # ผลรวมเริ่มต้น: f(a) + f(b)
    sum_val = func(a) + func(b)
    
    # วนลูปสำหรับจุดภายใน
    for i in range(1, n):
        x = a + i * h
        
        # กรณี Simpson 1/3:
        # ตำแหน่งคี่ (Odd) คูณ 4
        # ตำแหน่งคู่ (Even) คูณ 2
        if i % 2 == 1:
            sum_val += 4 * func(x)
        else:
            sum_val += 2 * func(x)
            
    # คูณด้วย h/3 ตามสูตร
    return (h / 3) * sum_val

# --- 2. แก้โจทย์ข้อ a ---
# โจทย์: Integral cos^3(x) dx จาก pi/2 ถึง 3pi/2
def func_a(x):
    return math.cos(x)**3

a_start = math.pi / 2
a_end = 3 * math.pi / 2
n = 100  # กำหนดจำนวนช่วง (ยิ่งเยอะยิ่งแม่น)

result_a = simpsons_one_third(func_a, a_start, a_end, n)
print(f"ข้อ a: ผลลัพธ์ = {result_a:.6f}")
# (ค่าทางทฤษฎีคือ -4/3 หรือประมาณ -1.3333...)


# --- 3. แก้โจทย์ข้อ b ---
# โจทย์: Integral 1 / (1 + 9x^2) dx จาก -1.2 ถึง 1.2
def func_b(x):
    return 1 / (1 + 9 * (x**2))

b_start = -1.2
b_end = 1.2
n = 100

result_b = simpsons_one_third(func_b, b_start, b_end, n)
print(f"ข้อ b: ผลลัพธ์ = {result_b:.6f}")
```

Simpson 3/8 Gemini

```python
import math

import math

# --- 1. นิยามฟังก์ชันสำหรับ Simpson's 3/8 Rule ---
def simpsons_three_eighths(func, a, b, n):
    """
    func: ฟังก์ชันที่ต้องการอินทิเกรต
    a: ขอบเขตล่าง
    b: ขอบเขตบน
    n: จำนวนช่วงย่อย (ต้องหารด้วย 3 ลงตัว)
    """
    if n % 3 != 0:
        return "Error: n ต้องหารด้วย 3 ลงตัว สำหรับ Simpson's 3/8 Rule"
    
    h = (b - a) / n
    
    # ผลรวมเริ่มต้น: f(a) + f(b) (ตัวแรกและตัวสุดท้ายคูณ 1)
    sum_val = func(a) + func(b)
    
    # วนลูปสำหรับจุดภายใน
    for i in range(1, n):
        x = a + i * h
        
        # เงื่อนไขตัวคูณของ Simpson 3/8:
        # ถ้าตำแหน่งหาร 3 ลงตัว (เช่น 3, 6, 9) ให้คูณ 2
        # นอกนั้นคูณ 3
        if i % 3 == 0:
            sum_val += 2 * func(x)
        else:
            sum_val += 3 * func(x)
            
    # คูณด้วย 3h/8 ตามสูตร
    return (3 * h / 8) * sum_val

# --- 2. แก้โจทย์ข้อ a ---
# โจทย์: Integral cos^3(x) dx จาก pi/2 ถึง 3pi/2
def func_a(x):
    return math.cos(x)**3

a_start = math.pi / 2
a_end = 3 * math.pi / 2
n_a = 99  # เลือก n ที่หารด้วย 3 ลงตัว (เช่น 99)

result_a = simpsons_three_eighths(func_a, a_start, a_end, n_a)
print(f"ข้อ a (Simpson 3/8): ผลลัพธ์ = {result_a:.6f}")


# --- 3. แก้โจทย์ข้อ b ---
# โจทย์: Integral 1 / (1 + 9x^2) dx จาก -1.2 ถึง 1.2
def func_b(x):
    return 1 / (1 + 9 * (x**2))

b_start = -1.2
b_end = 1.2
n_b = 99  # เลือก n ที่หารด้วย 3 ลงตัว

result_b = simpsons_three_eighths(func_b, b_start, b_end, n_b)
print(f"ข้อ b (Simpson 3/8): ผลลัพธ์ = {result_b:.6f}")
```

Simpson1/3 + Simpson3/8 Gemini

```python
import math

# 1. นิยามฟังก์ชันสำหรับโจทย์ข้อ a และ b
def func_a(x):
    return math.cos(x)**3

def func_b(x):
    return 1 / (1 + 9 * (x**2))

# 2. สร้างฟังก์ชัน Mixed Simpson's Rule (3/8 + 1/3)
def mixed_simpson(func, a, b, n):
    """
    คำนวณอินทิเกรตโดยใช้ Simpson's 3/8 ใน 3 ช่วงแรก
    และใช้ Simpson's 1/3 ในช่วงที่เหลือ
    เงื่อนไข: n ต้องเป็นเลขคี่ (Odd number) และ n >= 3
    """
    if n % 2 == 0 or n < 3:
        return "Error: n ต้องเป็นเลขคี่และมากกว่าหรือเท่ากับ 3 สำหรับวิธีนี้"

    h = (b - a) / n
    
    # --- ส่วนที่ 1: Simpson's 3/8 (3 ช่วงแรก: x0 ถึง x3) ---
    # จุด x0, x1, x2, x3
    x0 = a
    x1 = a + 1*h
    x2 = a + 2*h
    x3 = a + 3*h
    
    # สูตร Simp 3/8: (3h/8) * (f0 + 3f1 + 3f2 + f3)
    area_38 = (3 * h / 8) * (func(x0) + 3*func(x1) + 3*func(x2) + func(x3))
    
    # --- ส่วนที่ 2: Simpson's 1/3 (ช่วงที่เหลือ: x3 ถึง xn) ---
    rem_n = n - 3  # จำนวนช่วงที่เหลือ (จะเป็นเลขคู่เสมอ)
    
    area_13 = 0
    if rem_n > 0:
        # จุดเริ่มต้นของส่วนนี้คือ x3, จุดสิ้นสุดคือ b
        start_node = x3
        
        # ผลรวมหัวท้ายของส่วน 1/3 (คือ f(x3) + f(xn))
        sum_rem = func(start_node) + func(b)
        
        # วนลูปจุดภายในของส่วนที่เหลือ
        for i in range(1, rem_n):
            curr_x = start_node + i * h
            if i % 2 != 0:  # ตำแหน่งคี่ (ในส่วนย่อย) คูณ 4
                sum_rem += 4 * func(curr_x)
            else:           # ตำแหน่งคู่ (ในส่วนย่อย) คูณ 2
                sum_rem += 2 * func(curr_x)
        
        area_13 = (h / 3) * sum_rem
    
    return area_38 + area_13

# =================== การคำนวณ ===================

# กำหนดจำนวนช่วง n (ต้องเป็นเลขคี่ ตามโจทย์)
n_intervals = 99 

# --- ข้อ a ---
a_start = math.pi / 2
a_end = 3 * math.pi / 2
result_a = mixed_simpson(func_a, a_start, a_end, n_intervals)

print(f"ข้อ a (n={n_intervals}):")
print(f"ผลลัพธ์ = {result_a:.6f}")
print(f"ค่าทางทฤษฎี (-4/3) = -1.333333") 
print("-" * 30)

# --- ข้อ b ---
b_start = -1.2
b_end = 1.2
result_b = mixed_simpson(func_b, b_start, b_end, n_intervals)

print(f"ข้อ b (n={n_intervals}):")
print(f"ผลลัพธ์ = {result_b:.6f}")
```
